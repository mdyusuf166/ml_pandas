```python
import pandas as pd

# Sample data (wide-format)
data = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "Salary_2023": [70000, 60000, 50000],
    "Salary_2024": [75000, 62000, 52000],
    "Bonus_2023": [5000, 4000, 3000],
    "Bonus_2024": [6000, 4500, 3500]
})

print("মূল DataFrame (Wide Format):")
print(data)
```

```python
melted = pd.melt(
    data,
    id_vars=["EmployeeID", "Name"],  # এই কলামগুলো অপরিবর্তিত থাকবে
    value_vars=["Salary_2023", "Salary_2024"],  # যেগুলো unpivot করা হবে
    var_name="Year",  # নতুন কলামের নাম
    value_name="Salary"  # মানের কলামের নাম
)

print("Melted DataFrame (Salary):")
print(melted)
```

```python
import pandas as pd

# 1️⃣ Sample Employee Data
data = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Finance", "Finance"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Annual Salary": [70000, 65000, 50000, 48000, 90000, 85000],
    "City": ["Dhaka", "Dhaka", "Sylhet", "Sylhet", "CTG", "CTG"]
})

print("Original Data:")
print(data)

```

```python
# 2️⃣ Basic Pivot Table
pivot_basic = data.pivot_table(
    values="Annual Salary",
    index="Department",
    columns="Gender"
)
print("\nBasic Pivot Table:")
print(pivot_basic)
```
